
# What is Pandas?

Pandas is a Python library for data manipulation and analysis.
It provides two main data structures:

- Series – a one-dimensional labeled array.

- DataFrame – a two-dimensional table (rows and columns), similar to an Excel sheet or SQL table.

Pandas makes it easy to:

- Load data from files (CSV, Excel, SQL, JSON, etc.)
- Clean, filter, and transform data
- Perform statistical analysis
- Save results back to different file formats

# Reading a Dataset from the Internet

Pandas can read data directly from a URL.
Here we’ll use an open dataset called Penguins, which contains biological measurements for three penguin species.




```python
import pandas as pd
# Define the dataset URL
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/penguins.csv"

# Read the CSV file from the internet into a DataFrame
df = pd.read_csv(url)

# Display the first few rows
df.head()


# Define the local path where we want to save the file 
output_path = "../data/penguins_local.csv"

# Save the DataFrame as a CSV file
df.to_csv(output_path, index=False)

# Confirm the file has been saved
print(f"File saved locally as: {output_path}")
```

# Understanding What Happened

The `pd.read_csv()` function:

- Downloads the data (if a URL is given)
- Parses the comma-separated values (CSV format)
- Loads them into a DataFrame object called df

Saving the Data Locally

After we read the data, we can save it as a local CSV file.
This is useful when you want to reuse the data later without downloading it again.
