# Cleaning your_data
Before we can visualize the data, we need to make sure it’s clean and consistent.
This dataset is relatively clean, but it contains a few missing values that we’ll handle.

1. Check for Missing Values
   Let’s start by checking how many missing values we have in each column. use `df.isnull().sum()`

2. Decide How to Handle Them
  There are two common approaches:
  Drop rows with missing values (if they are few).
  Fill them with reasonable defaults (e.g., mean, median, or a placeholder).

3. Confirm Dataset Shape After Cleaning
   Let’s verify how many rows remain after removing missing data.

You’ll notice that only a small number of rows were removed, 
this ensures our dataset remains large enough for meaningful analysis.





```python
import pandas as pd
# Read the CSV file 
df = pd.read_csv("../data/penguins_local.csv")

# Count missing (NaN) values in each column
df.isnull().sum()
```

```python
# Since our dataset has very few missing values, we’ll drop them to keep the data simple and clean.
# Remove rows with any missing values
df_clean = df.dropna()

# Check again to confirm
df_clean.isnull().sum()
```

```python
# Let’s verify how many rows remain after removing missing data.
print("Original shape:", df.shape)
print("Cleaned shape:", df_clean.shape)
```
