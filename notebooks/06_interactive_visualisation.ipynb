## Introduction

In this notebook, we explore **interactive data visualization** using the **Holoviews** and **hvplot** libraries.
While libraries like **Seaborn** are excellent for creating *static* and publication-ready plots, Holoviews focuses on *interactive exploration* directly inside Jupyter notebooks.

With Holoviews (built on top of **Bokeh**), you can zoom, pan, hover, and dynamically select variables — without writing additional plotting code for each case.
This makes it ideal for **data investigation and dashboard-style analytics**, where you want to quickly explore relationships between different features instead of generating many separate static charts.


### Quick Interactive Visualization with hvPlot

The following example demonstrates how to create an **interactive scatter plot** using the `hvPlot` library, which provides an intuitive interface for data exploration built on top of **Holoviews** and **Bokeh**.

Unlike Seaborn or Matplotlib, `hvPlot` allows users to instantly add interactivity, including:
- zooming
- panning
- hover tooltips 

Here, we visualize the relationship between penguin body mass and flipper length, coloring the points by species.
The plot is interactive by default and can be explored directly inside JupyterLab without any additional setup or callbacks.


```python
import pandas as pd
import hvplot.pandas  # loads hvplot extension automatically

# Load data
df = pd.read_csv("../data/penguins_local.csv")

# Simple interactive scatter plot
df.hvplot.scatter(
    x="flipper_length_mm",
    y="body_mass_g",
    by="species",
    height=400,
    width=600,
    alpha=0.6,
    legend="top_left",
    title="Penguin Body Mass vs Flipper Length"
)
```

## Interactive Data Exploration with Panel and hvPlot

In this section, we use **Panel** together with **hvPlot** to build an interactive data exploration tool directly inside JupyterLab.
`hvPlot` extends pandas’ plotting interface to automatically create interactive visualizations using **Bokeh** as a backend, while **Panel** provides a lightweight framework for adding UI controls such as dropdown menus and sliders.

Here, we load the penguin dataset and create two dropdown selectors that let the user dynamically choose which numeric columns to plot on the X and Y axes.
As the user changes the selections, the scatter plot updates automatically to show the new relationship — making it easy to explore correlations and patterns in the data without writing new plotting code each time.


```python
import panel as pn
import hvplot.pandas
pn.extension('tabulator')

df = pd.read_csv("../data/penguins_local.csv")
numeric_cols = list(df.select_dtypes(include='number').columns)

x = pn.widgets.Select(name='X-axis', options=numeric_cols, value='flipper_length_mm')
y = pn.widgets.Select(name='Y-axis', options=numeric_cols, value='body_mass_g')

@pn.depends(x, y)
def interactive_plot(x, y):
    return df.hvplot.scatter(x=x, y=y, by="species", alpha=0.6, height=400, width=600)

pn.Column(pn.Row(x, y), interactive_plot).servable()
```
