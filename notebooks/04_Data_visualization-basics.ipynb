#  Basic_Data_Visualization_with_Seaborn

### Introduction
**Data visualization** helps us understand relationships and patterns in data that might not be obvious from tables or numbers alone.  
We’ll use **Seaborn**, a powerful visualization library built on top of Matplotlib, which works seamlessly with Pandas DataFrames.

Before we start, let’s import the necessary libraries:


```python
import pandas as pd
# Read the CSV file 
df = pd.read_csv("../data/penguins_local.csv")

# Count missing (NaN) values in each column
df.isnull().sum()
import seaborn as sns
import matplotlib.pyplot as plt

# Use a clean style
sns.set_theme(style="whitegrid")

# We’ll use the cleaned penguins dataset
df = df.copy()
```

## 1. Box Plot – Distribution by Category

Box plots show the distribution of a numeric variable across categories, including the median, quartiles, and outliers.


```python
# Boxplot: body mass by species
sns.boxplot(x="species", y="body_mass_g", data=df)
plt.title("Body Mass Distribution by Penguin Species")
plt.show()
```

### 2. Bar Plot – Comparing Averages

Bar plots display the mean (or another statistic) of a numeric variable across categories.


```python
# Barplot: average flipper length by species
sns.barplot(
    x="species",
    y="flipper_length_mm",
    data=df,
    estimator='mean',
    errorbar='sd'   # replaces ci='sd'
)
plt.title("Average Flipper Length by Species")
plt.show()

```

### 3. Histogram – Distribution of a Single Variable
Histograms show how data points are distributed across different bins.

```python
# Histogram: body mass distribution
sns.histplot(df["body_mass_g"], bins=20, kde=True)
plt.title("Distribution of Penguin Body Mass")
plt.xlabel("Body Mass (g)")
plt.ylabel("Count")
plt.show()
```

### 4. Scatter Plot – Relationship Between Two Variables
Scatter plots show how two continuous variables relate to each other.

```python
# Scatterplot: flipper length vs body mass
sns.scatterplot(x="flipper_length_mm", y="body_mass_g", data=df)
plt.title("Flipper Length vs Body Mass")
plt.show()
```

### 5. Enhanced Scatter Plot – Adding a Third Dimension (Color and Size)

We can add more dimensions to a scatter plot by encoding variables as **color** and **size**.


```python
# Scatterplot with hue and size
sns.scatterplot(
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species",
    size="bill_length_mm",
    sizes=(40, 200),
    data=df
)
plt.title("Flipper Length vs Body Mass, Colored by Species and Sized by Bill Length")
plt.show()
```

### 6. Pair Plot – Exploring Multiple Relationships

Pair plots show relationships between all pairs of numerical features at once.

```python
# Pairplot: automatic comparison of all numeric columns
sns.pairplot(df, hue="species")
plt.suptitle("Pairwise Relationships Between Penguin Measurements", y=1.02)
plt.show()
```
