##  About the Dataset: *Palmer Penguins* 🐧

### Overview

The **Palmer Penguins** dataset is a modern alternative to the classic *Iris* dataset.
It provides real-world biological data collected from penguin species living on the **Palmer Archipelago**, near Antarctica.
The data were gathered by **Dr. Kristen Gorman** and the **Palmer Station Long-Term Ecological Research (LTER)** program.

The dataset was created to offer a more interesting, realistic, and inclusive example for teaching data analysis and visualization, compared to the older and smaller *Iris* dataset.

Original source:

> Gorman, K.B., Williams, T.D., & Fraser, W.R. (2014).
> *Ecological Sexual Dimorphism and Environmental Variability within a Community of Antarctic Penguins (Genus Pygoscelis).*
> PLoS ONE 9(3): e90081.
> [https://doi.org/10.1371/journal.pone.0090081](https://doi.org/10.1371/journal.pone.0090081)

### Variables Description

| Column                | Type          | Description                                                                                                        |
| :-------------------- | :------------ | :----------------------------------------------------------------------------------------------------------------- |
| **species**           | *categorical* | The penguin species: *Adelie*, *Chinstrap*, or *Gentoo*.                                                           |
| **island**            | *categorical* | The island where the penguin was observed: *Biscoe*, *Dream*, or *Torgersen*.                                      |
| **bill_length_mm**    | *numeric*     | The length of the penguin’s bill (beak) in millimeters.                                                            |
| **bill_depth_mm**     | *numeric*     | The depth (height) of the bill in millimeters.                                                                     |
| **flipper_length_mm** | *numeric*     | The length of the penguin’s flipper (wing) in millimeters.                                                         |
| **body_mass_g**       | *numeric*     | The body mass of the penguin in grams.                                                                             |
| **sex**               | *categorical* | The biological sex of the penguin (*Male*, *Female*). Some values are missing because not all penguins were sexed. |
| **year**              | *numeric*     | The year of data collection (2007, 2008, or 2009).                                                                 |

---

### Why This Dataset Is Useful

* Clean and small enough for quick experiments.
* Includes both **numerical** and **categorical** data.
* Great for practicing **data cleaning**, **grouping**, **aggregation**, and **visualization**.
* Used widely in education and tutorials for **Pandas**, **Seaborn**, and **scikit-learn**.
