# Advanced Data Visualization with Seaborn

### Introduction
Now that we’ve covered basic charts, let’s move on to **more advanced visualizations**.  
These examples will show how to display multi-dimensional data, relationships, correlations, and distributions in more complex but insightful ways.

### 1. Heatmap – Visualizing Correlations

A heatmap is a graphical representation of data where values are represented by colors.  
It’s often used to visualize **correlation matrices**.


```python
import pandas as pd


import seaborn as sns
import matplotlib.pyplot as plt
# Read the CSV file 
df_clean = pd.read_csv("../data/penguins_local.csv")
# Use a clean style
sns.set_theme(style="whitegrid")

# We’ll use the cleaned penguins dataset
df = df_clean.copy()
```

```python

# Compute correlation between numeric columns
corr = df_clean.corr(numeric_only=True)

# Create the heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap of Penguin Measurements")
plt.show()
```

### 2. Violin Plot – Combining Distribution and Summary Statistics

Violin plots combine a box plot with a kernel density estimate (KDE)  
to show both **the spread and the shape** of the data distribution.



```python
sns.violinplot(x="species", y="body_mass_g", data=df_clean, inner="quartile")
plt.title("Body Mass Distribution by Species (Violin Plot)")
plt.show()
```

## 3. Swarm Plot – Showing Every Observation

A swarm plot displays every single data point, spaced out to avoid overlap.  
It’s useful to show individual variability within categories.


```python
sns.swarmplot(x="species", y="flipper_length_mm", data=df_clean, size=4)
plt.title("Flipper Length Distribution by Species (Smaller Points)")
plt.show()
```

### 4. FacetGrid – Multiple Plots by Category

FacetGrid allows you to create **multiple subplots** for different categories automatically.


```python
g = sns.FacetGrid(df_clean, col="species")
g.map(sns.histplot, "body_mass_g", bins=15, kde=True)
g.fig.suptitle("Body Mass Distribution by Species", y=1.05)
plt.show()
```

### 5. Joint Plot – Relationship + Distributions

A joint plot shows both a scatter plot and histograms for the two variables.  
It’s ideal for exploring **bivariate relationships**.

```python
sns.jointplot(
    data=df_clean,
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species",
    kind="scatter"
)
plt.suptitle("Joint Plot: Flipper Length vs Body Mass", y=1.02)
plt.show()
```

### 6. PairGrid – Full Control Over Pairwise Relationships

`PairGrid` is a more customizable version of `pairplot()`.  
You can choose different types of plots for the upper, lower, and diagonal parts of the grid.


```python
g = sns.PairGrid(df_clean, hue="species")
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot, fill=True)
g.map_diag(sns.histplot, kde=True)
g.add_legend()
plt.suptitle("Advanced PairGrid Visualization", y=1.02)
plt.show()
```

### 7. Regression Plot – Trend Lines

You can easily visualize trends or linear relationships using regression plots.


```python
sns.lmplot(
    x="flipper_length_mm",
    y="body_mass_g",
    hue="species",
    data=df_clean,
    height=6
)
plt.title("Regression Line: Flipper Length vs Body Mass by Species")
plt.show()
```

### 8. Categorical Comparison – Multiple Variables

You can visualize two categorical variables with color (hue) and position (x-axis).


```python
sns.catplot(
    data=df_clean,
    x="island",
    y="body_mass_g",
    hue="species",
    kind="bar"
)
plt.title("Average Body Mass per Island by Species")
plt.show()
```


### Summary

In this notebook, we explored:
- Correlation heatmaps  
- Violin and swarm plots  
- Multi-panel plots with FacetGrid  
- Joint and PairGrid plots  
- Regression and categorical comparison plots  

Each of these tools helps reveal **patterns, clusters, or correlations** in complex datasets.
