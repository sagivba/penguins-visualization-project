# Exploring the Dataset

Before visualizing or analyzing data, it’s important to explore it — to understand its structure, columns, data types, and potential issues (like missing values).

1. Checking the Basic Structure
   `df.info()` gives an overview of the dataset — number of rows, columns, data types, and missing values.
2. Viewing Basic Statistics
   `df.describe()` provides descriptive statistics (mean, standard deviation, min, max, etc.) for numeric columns.
    You can also include non-numeric columns by using: `df.describe(include='all')`
3. Checking the Shape and Columns
   It’s useful to know how many rows and columns you have, and what each column represents.
4. Checking for Missing Values
   Missing data is common. Pandas lets you easily identify it.
5. Previewing the Data
  It’s always a good idea to manually inspect a few records. use `df.head()` and `df.tail()`
6. Quick Summary of Categorical Columns
   To better understand the categorical variables (like species or island), use value_counts().
      




```python
import pandas as pd
# Read the CSV file 
df = pd.read_csv("../data/penguins_local.csv")




```

```python
# Display general information about the dataset
df.info()
```

```python
# Display basic statistics for numeric columns
df.describe()
```

```python
# Include all columns (numeric + categorical)
df.describe(include='all')
```

```python
# Number of rows and columns
print("Dataset shape:", df.shape)

# List of all column names
print("\nColumns:")
print(df.columns.tolist())
```

```python
# Count missing values in each column
df.isnull().sum()
```

```python
# Show the first 5 rows
df.head()
```

```python
# Show the last 5 rows
df.tail()
```

```python
# Count unique values in categorical columns
print("Species distribution:")
print(df['species'].value_counts())

print("\nIsland distribution:")
print(df['island'].value_counts())
```

# Summarizing All Columns

This code gives a complete overview of all columns — their data types, number of unique values, missing values, and basic statistics.


```python

# General dataset summary
print("=== DataFrame Overview ===")
df.info()

# Summary statistics for all columns (numeric + categorical)
summary = df.describe(include='all').transpose()
summary['missing_values'] = df.isnull().sum()
summary['unique_values'] = df.nunique()

# Display summary
pd.set_option('future.no_silent_downcasting', True)
summary[['count', 'unique_values', 'missing_values', 'mean', 'min', 'max']].fillna('-')
```
